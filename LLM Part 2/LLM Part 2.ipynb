## **Generalized overview of NLP**
Steps to take: <br>**Text pre-processing -> Text representation -> Pre-training -> Fine-tuning -> Advanced Fine-tuning**

---
## 1. Text pre-processing:
***1.1. Tokenization:*** Splits text individual words = **tokens**<br>
***1.2. Stop word removal:*** Words that don't add meaning are eliminated (with, and, is, ...)<br>
***1.3. Lemmatization:*** Groups slightly different words with similar meanings and maps them to base form (talking, talked, talk --> talk)<br>

---
## 2. Text representation:
Text into numerical form through **bag of words** or **word embeddings**<br><br>
***Bag of words:*** text into a matrix of word counts (1: word presence, 0: absence)<br>
### Limitations: 
* doesn't capture order or context --> leads to incorrect interpretations
* doesn't capture semantics between words (cat chased mouse = mouse chased cat)<br>

***Word embeddings:*** capture semantic meanings as numbers<br>

### Largeness challenges:
Large amounts of data, efficient model training methods, powerful computers
<br>***Computing power:*** LLMs require 100,000's of CPUs and 10,000's of GPUs for **pre-training**<br>
**-> Fine-tuning computes with 1-2 CPU and GPU**<br>
***Efficient model training:*** Pre-training in itself takes months<br>
**-> Fine-tuning takes hours to days**<br>
***Data avilability:*** Pre-training needs high-quality data = ~10² GB of text (<=> +1M books) to learn complexities of language<br>
**-> Fine-tuning requires ~1 GB**<br>
<br>
*--> Fine-tuning adapts a pre-trained models and addresses those challenges*<br>
*=! Pre-training is not optimized for specific tasks and can be fine-tuned for a specific problem*<br>
### Learning techniques: (what if there is little to no labeled data) 
### --> N-shot learning
**Transfer learning:** learning from one task and transferring knowledge to a related task
<br><br>
**Zero-shot learning:**<br>
* no explicit training
* uses language understanding and context
* generalizes without any prior examples<br>

**Few-shot learning:**<br>
* learn a new task with a few examples
* uses prior knowledge to answer anew question(s)<br>

**Multi-shot learning:**<br>
* more examples than few-shot
* uses previous tasks and new examples<br>
