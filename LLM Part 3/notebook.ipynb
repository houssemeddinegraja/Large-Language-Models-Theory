# Chapter 3: Training Methodology and Techniques
## Building blocks to train LLMs
## Pre-training
LLMs are typically trained using a technique called <b>generative pre-training</b> which involves:
* providing a dataset of text tokens as input
* training the model to predict the tokens of that dataset
<br>
*Styles of generative pre-training:* **Next word prediction** and **masked language modeling**<br>

### Next word prediciton:
* Supervized learning technique - model trained on input-output pairs
* Predicts next word and generates coherent text
* Captures dependencies between words
* **More examples = better prediction**
<br>
### Masked language modeling:
* Hides (masks) a selective word - model trained to predict the masked word
* Based on learnings from training data
---
## Introducing the TRANSFORMER
### Revolutionized language modeling
* Relationship between words
* **Components**: Pre-processing -> Positional Encoding -> Encoders -> Decoders

### 1. Text pre-processing and representation:
* Tokenization, stop word removal, lemmatization
* Word embedding
### 2. Positional encoding:
* Information on the position of each word
* Understand distant words
### 3. Encoders:
* Directs attention to specific words and relationships = *Attention mechanism*
* Process specific features = *Neural network*
### 4. Decoders:
* Also includes attention and neural networds
* Generates the output
<br>
## Long-range dependencies challenge
-> Overcomed thanks to attention mechanism (focus on different parts of the input)
## Sequentiality challenge (traditional LMs process one word at a time)
-> Transformers process multiple parts simultaneously, and faster

---
## Attention mechanisms
They understand complex structures + focus on important words/relevant content
## Self-attention:
* Weighs the importance of each word
* Captures long-range dependencies
## Multi-head attention:
* Next level of self-attention
* Splits input into multiple heads with each head focusing on different aspects
### Example: at a party
**Self-attention**
* Focus on each person's words
* Evaluate and compare their relevance
* Weigh each speaker's input
* Combines for a comprehensive understanding<br>

**Multi-head attention**
* Split attention into "multiple" channels
* Focus on different aspects of conversation
* Speaker's emotions, primary topic, and related side-topics
* Process each aspect and merge
---
## Advanced fine-tuning
### --> Reinforcement Learning through Human Feedback (RLHF)
### Why ?
* General-purpose training data **lacks quality** --> Noise, errors, reduced accuracy
* **The start of the need of fine-tuning**:<br>
  --> Pre-training: learns underlying language patterns, doesn't capture context-specific complexities<br>
  --> Fine-tuning: quality labeled data improves performance<br>
  --> **Enter RLHF:** **Human feedback** = model output reviewed by human, and updated based on the feedback<br><br>
**Steps**:
1. Receives a prompt - generalizes multiple responses
2. Human expert checks these responses - ranks them based on quality (accuracy, relevance, coherence)
3. Learns from expert's ranking to align the responses with the expert's prefernces in the future
   And so on...
